```python
import json

# Read the original JSON file
with open('demographic/real.json', 'r', encoding='utf-8') as f:
    original_data = json.load(f)
# Transform the data (only keep entries with answerid 1, 2, or 3)
transformed_data = []
for entry in original_data:
    answerid = entry["questions"]["Q4"]["answerid"]
    if answerid in [1, 2, 3]:
        new_entry = {
            "userstudyid": entry["user_study_id"],
            "llm_answer": answerid
        }
        transformed_data.append(new_entry)

# Save the new JSON file
with open('transformed_data.json', 'w', encoding='utf-8') as f:
    json.dump(transformed_data, f, ensure_ascii=False, indent=2)

print(f"Transformation complete! Processed {len(transformed_data)} entries.")
print("Output saved to: transformed_data.json")
```
